在 LangChain 的 Agent 执行循环中，比如 “模型调用前”、“模型调用后”、“工具调用前后” 设置一些钩子
（hooks），让你在不改 Agent 主体逻辑的情况下实现策略与治理。

内置中间件：LangChain实现并提供的
